## 오차 역전파 그래프

```python
x, y, z = 100, 200, 0.5

q = x+y; f = q*z
print(f'result of front propagation: q={q}, f={f}')
```

```python
dfdz = q
dfdq = z
dqdx = 1.0
dqdy = 1.0

dfdx = dfdq * dqdx
dfdy = dfdq * dqdy

print(f'df/df={dfdx}, df/dy={dfdy}, df/dz={dfdz}')
```

## 시그모이드 함수

```python
import numpy as np

def sigmoid(x):
    return 1 / (1+np.exp(-x))

def sigmoid_differential(x):
    return sigmoid(x)*(1-sigmoid(x))

import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)
plt.plot(x, sigmoid(x))
plt.title('sigmoid')
display(plt.show())

plt.plot(x, sigmoid_differential(x))
plt.ylim(0, 1)
plt.title('sigmoid(differented)')
plt.show()
```

```python
import math

n0 = 1.0
n1 = n0 * -1.0
n2 = math.e ** (n1)
n3 = n2+1
n4 = 1/n3

print(f'{n0}, {n1}, {n2:.3f}, {n3:.3f}, {n4:.3f}')
```

## ReLU

```python
#시그모이드 함수의 기울기 소실
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


for x in range(-2, 3):
    print(f'sigmoid{x}: {sigmoid(x)}')
```

```python
# ReLU funciton
def ReLU(x):
    return np.maximum(0, x)

for x in range(-2, 3):
    print(f'relu({x}): {ReLU(x)}')
```
