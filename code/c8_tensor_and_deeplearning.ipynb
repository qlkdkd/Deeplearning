### 완전 연결층

```python
from tensorflow import keras

d1 = keras.layers.Dense(
    32, activation='relu', input_shape=(4,),
    name='layer1')

d2 = keras.layers.Dense(30, activation='relu')

d3 = keras.layers.Dense(3, activation='softmax')

s_model = keras.Sequential([d1, d2, d3])
s_model.summary()
```
